```python
import csv
with open ('student.csv',encoding='utf-8',newline='') as file:
    reader_dict=list(csv.DictReader(file))
    print(reader_dict)
    for item in reader_dict:
        print(item)
    new_list=[]
    for item in reader_dict:
        if int(item['age']) > 20:
            print(item)
            new_list.append(item)
print(new_list)
```
